## Itô's Lemma and the Black-Scholes Derivation

---

#### Overview

Itô's Lemma is the stochastic calculus analogue of the chain rule. It describes how a smooth function of a stochastic process evolves over time. The derivation begins with the Taylor Series and carefully tracks which terms survive when the underlying process is driven by a Wiener process — the key surprise being that second-order terms in $dW$ do not vanish.

### 1. Taylor Series Review

The Taylor Series expresses any infinitely differentiable function as an infinite-degree polynomial expansion around a chosen point $a$:

### $$f(x) = \frac{f(a)}{0!} + \frac{(x-a)f'(a)}{1!} + \frac{(x-a)^2 f''(a)}{2!} + \cdots + \frac{(x-a)^n f^{(n)}(a)}{n!}$$

#### Example: $f(x) = \sqrt{x}$ expanded around $a = 9$

### $$f(x) = x^{1/2} \qquad f(9) = 3$$

### $$f'(x) = \frac{1}{2}x^{-1/2} = \frac{1}{2\sqrt{x}} \qquad f'(9) = \frac{1}{6}$$

### $$f''(x) = -\frac{1}{4}x^{-3/2} = -\frac{1}{4(\sqrt{x})^3} \qquad f''(9) = -\frac{1}{108}$$

### $$\Rightarrow \quad f(x) \approx 3 + \frac{(x-9)}{6} - \frac{(x-9)^2}{216}$$

#### Chain Rule (Ordinary Calculus)

For a composition $f(h(x))$:

### $$\frac{d}{dx}f(h(x)) = f'(h(x)) \cdot h'(x)$$

This is the rule that needs to be modified for stochastic processes.

### 2. Why Ordinary Calculus Breaks Down

In ordinary calculus, for a smooth deterministic function $x(t)$:

### $$dx \approx x'(t)\, dt \quad \Rightarrow \quad (dx)^2 \approx \bigl(x'(t)\bigr)^2 (dt)^2$$

Since $(dt)^2$ is second-order in $dt$, it vanishes faster than $dt$ itself as $dt \to 0$. The Taylor expansion for $f(t, x)$ therefore keeps only first-order terms, and the standard chain rule follows.

###
| $dt$ | first-order $dt$ | second-order $(dt)^2$ | ratio $(dt)^2 / dt$ |
|---|---|---|---|
| $10^{-2}$ | $0.01$ | $0.0001$ | $0.01$ |
| $10^{-4}$ | $0.0001$ | $10^{-8}$ | $0.0001$ |
| $10^{-6}$ | $0.000001$ | $10^{-12}$ | $0.000001$ |

Relative to the first-order term, the second-order column is effectively zero — which is exactly the license we use to drop $(dx)^2$. It works because a smooth path's squared increments, summed up, also vanish:

### $$\sum_i (\Delta x_i)^2 \approx (x')^2\, t \cdot \Delta t \;\xrightarrow{\Delta t \to 0}\; 0.$$


### 3. Itô Processes and Wiener Process Properties

#### Itô Process

A stochastic process $X(t)$ following an Itô process satisfies:

### $$dX(t) = \underbrace{a(t,X)}_{\text{drift}}\,dt + \underbrace{b(t,X)}_{\text{diffusion}}\,dW(t)$$

where $a(t, X)$ is the drift coefficient, $b(t, X)$ is the diffusion coefficient, and $W(t)$ is a standard Wiener process (Brownian motion).

#### Itô Multiplication Rules

These follow from the quadratic variation properties of Brownian motion:

### $$dW^2 = dt$$
### $$dt\,dW = 0$$
### $$dt^2 = 0$$

The crucial identity is $(dW)^2 = dt$. Intuitively, Brownian motion accumulates quadratic variation at a constant rate — unlike a smooth path whose squared increments are negligible. This is what makes second-order terms survive in the stochastic Taylor expansion and distinguishes Itô calculus from ordinary calculus.

### 4. Deriving Itô's Lemma

Let $f(t, X)$ be twice continuously differentiable. Expand using the multivariable Taylor series:

### $$df(t, X) = \frac{\partial f}{\partial t}\,dt + \frac{\partial f}{\partial x}\,dX + \frac{1}{2}\left(\frac{\partial^2 f}{\partial t^2}(dt)^2 + \frac{\partial^2 f}{\partial x^2}(dX)^2\right) + \cdots$$

Substitute $dX = a\,dt + b\,dW$:

### $$df(t,X) = \frac{\partial f}{\partial t}\,dt + \frac{\partial f}{\partial x}\bigl(a\,dt + b\,dW\bigr) + \frac{1}{2}\frac{\partial^2 f}{\partial x^2}(dX)^2 + \cdots$$

Expand $(dX)^2$ and apply the Itô multiplication rules:

### $$ (dX)^2 = a^2\underbrace{(dt)^2}_{=\,0} + 2ab\underbrace{\,dt\,dW}_{=\,0} + b^2\underbrace{(dW)^2}_{=\,dt} \;=\; b^2\,dt $$

Substituting back:

### $$df(t,X) = \frac{\partial f}{\partial t}\,dt + \frac{\partial f}{\partial x}\bigl(a\,dt + b\,dW\bigr) + \frac{1}{2}\frac{\partial^2 f}{\partial x^2}\,b^2\,dt$$

### 5. Itô's Lemma — Final Form

Collecting all $dt$ terms and the $dW$ term:

### $$df(t,X) = \left(\frac{\partial f}{\partial t} + a(t,X)\frac{\partial f}{\partial x} + \frac{1}{2}b^2(t,X)\frac{\partial^2 f}{\partial x^2}\right)dt + b(t,X)\frac{\partial f}{\partial x}\,dW(t)$$

#### Comparison to the Ordinary Chain Rule

The standard chain rule for a deterministic process gives:

$$df = \frac{\partial f}{\partial t}\,dt + \frac{\partial f}{\partial x}\,dX$$

Itô's Lemma adds the correction term $\dfrac{1}{2}b^2\dfrac{\partial^2 f}{\partial x^2}\,dt$, which arises entirely from the non-zero quadratic variation of Brownian motion. In ordinary calculus this term is zero; in stochastic calculus it is not.

### Black-Scholes Derivation from Itô's Lemma

#### Overview

With Itô's Lemma in hand, we can derive the Black-Scholes PDE. The argument has three steps: model the stock price as a geometric Brownian motion, apply Itô's Lemma to the option price, then construct a portfolio that eliminates all randomness. A risk-free portfolio must earn exactly the risk-free rate — this no-arbitrage condition yields the PDE.

### 6. Stock Price Dynamics — Geometric Brownian Motion

We model the stock price $S(t)$ as a geometric Brownian motion (GBM):

$$dS = \mu S\,dt + \sigma S\,dW$$

This is an Itô process with drift $a = \mu S$ and diffusion $b = \sigma S$, where:

- $\mu$ — the expected continuously compounded return of the stock (drift)
- $\sigma$ — the volatility of the stock (standard deviation of returns per unit time)
- $W(t)$ — a standard Wiener process

GBM is chosen because it keeps $S$ strictly positive (no negative prices) and produces log-normally distributed returns, which matches the empirical behaviour of equity prices reasonably well.

Note that $\mu$ does not appear in the final Black-Scholes formula — it gets eliminated when we impose the no-arbitrage condition below.

### 7. Applying Itô's Lemma to the Option Price

Let $V(t, S)$ be the price of an option (or any derivative) written on $S$. We treat $V$ as a function of time $t$ and the stock price $S(t)$, and apply Itô's Lemma with $a = \mu S$ and $b = \sigma S$:

$$dV = \frac{\partial V}{\partial t}\,dt + \frac{\partial V}{\partial S}\,dS + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2}\,dt$$

Substituting $dS = \mu S\,dt + \sigma S\,dW$:

$$dV = \left(\frac{\partial V}{\partial t} + \mu S \frac{\partial V}{\partial S} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2}\right)dt + \sigma S \frac{\partial V}{\partial S}\,dW$$

The option price contains both a deterministic component (the $dt$ terms) and a stochastic component (the $dW$ term). The stochastic term is what we will eliminate next.

### 8. Delta Hedging — Eliminating the Stochastic Term

Construct a portfolio $\Pi$ consisting of one option and a short position of $\Delta$ shares:

$$\Pi = V - \Delta S$$

The change in portfolio value over an infinitesimal time step is:

$$d\Pi = dV - \Delta\,dS$$

Substituting the expressions for $dV$ and $dS$:

$$d\Pi = \left(\frac{\partial V}{\partial t} + \mu S \frac{\partial V}{\partial S} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2}\right)dt + \sigma S \frac{\partial V}{\partial S}\,dW - \Delta\left(\mu S\,dt + \sigma S\,dW\right)$$

Choose $\Delta = \dfrac{\partial V}{\partial S}$ (the option's delta). The $dW$ terms cancel exactly:

$$\sigma S \frac{\partial V}{\partial S}\,dW - \frac{\partial V}{\partial S}\cdot \sigma S\,dW = 0$$

What remains is entirely deterministic:

$$d\Pi = \left(\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2}\right)dt$$

This is the delta-hedged portfolio. By choosing $\Delta = \partial V / \partial S$, we have instantaneously eliminated all exposure to the random Brownian motion $W$, leaving a portfolio that is risk-free over the interval $dt$.

### 9. No-Arbitrage Condition and the Risk-Free Rate

Since $d\Pi$ contains no $dW$ term, the portfolio is instantaneously risk-free. By the no-arbitrage principle, a risk-free portfolio must earn exactly the risk-free rate $r$. Any other return would create an arbitrage opportunity:

- If $d\Pi > r\Pi\,dt$: borrow at $r$, invest in the portfolio, and earn a riskless profit.
- If $d\Pi < r\Pi\,dt$: sell the portfolio short and invest proceeds at $r$.

Therefore:

$$d\Pi = r\Pi\,dt$$

Substituting $\Pi = V - \dfrac{\partial V}{\partial S}S$:

$$d\Pi = r\left(V - \frac{\partial V}{\partial S}S\right)dt$$

### 10. The Black-Scholes PDE

Equating the two expressions for $d\Pi$:

$$\left(\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2}\right)dt = r\left(V - \frac{\partial V}{\partial S}S\right)dt$$

Dividing through by $dt$:

$$\boxed{\frac{\partial V}{\partial t} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + rS\frac{\partial V}{\partial S} - rV = 0}$$

This is the Black-Scholes PDE. Any derivative whose price $V(t, S)$ can be written as a smooth function of $t$ and $S$ must satisfy this equation, subject to its terminal payoff condition.

#### Boundary Conditions for a European Call

For a European call option with strike $K$ and expiry $T$, the terminal condition is:

$$V(T, S) = \max(S - K,\; 0)$$

with boundary conditions $V(t, 0) = 0$ and $V(t, S) \to S - Ke^{-r(T-t)}$ as $S \to \infty$.

#### Key Observations

The stock's expected return $\mu$ does not appear anywhere in the PDE. This is not an accident — when we eliminated $dW$ by delta hedging, we also eliminated $\mu$ (it appeared only inside the $dW$ coefficient). The PDE is therefore the same for all investors regardless of their views on the stock's drift, which is why option prices can be agreed upon in a market even when investors disagree about expected returns.

The parameters that do appear are $r$ (the risk-free rate), $\sigma$ (volatility), $S$ (current stock price), and $t$ (time). Solving the PDE with the terminal payoff as a boundary condition yields the Black-Scholes formula.
